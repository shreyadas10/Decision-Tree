1.  What is a Decision Tree, and how does it work	?

-> A decision tree is a flowchart-like structure that models decisions and their potential outcomes. It's a way to visualize and analyze different options to make a decision, often used in business planning, machine learning, and decision analysis.

2. What are impurity measures in Decision Trees	?

-> In decision trees, impurity measures quantify how "mixed" the classes are within a node. Common metrics include Gini impurity and entropy. These measures guide the splitting process, aiming to reduce impurity at each node and create a more homogenous distribution of classes in the resulting child nodes.


3. What is the mathematical formula for Gini Impurity	?

-> The mathematical formula for Gini Impurity measures the impurity of a set of items by determining how likely it is that a randomly chosen item from the set will be mislabeled if the set is randomly labeled. In essence, it quantifies how mixed the classes are within the set.

-> Gini = 1 - Σ (p(i)^2)

4. What is the mathematical formula for Entropy	?

-> Entropy measures how mixed or pure the dataset is. Lower entropy means purer (more similar) classes.

-> The mathematical formula is :  H(D) = - ∑ (p * log2 p)




5. What is Information Gain, and how is it used in Decision Trees	?

->  Information gain, measures the reduction in entropy (uncertainty) achieved by splitting a dataset based on a particular feature. It helps determine the most informative features for building a decision tree by identifying the splits that provide the most clarity in separating different classes.

6. What is the difference between Gini Impurity and Entropy	?

-> Gini impurity and entropy are both measures of impurity or randomness within a dataset, commonly used in decision tree algorithms to determine the best split for a node. While both aim to quantify uncertainty, they differ in their calculation and implications. Gini impurity is simpler to compute and generally faster, while entropy provides a more comprehensive view of data distribution, potentially leading to slightly more balanced trees but with a higher computational cost.

7. What is the mathematical explanation behind Decision Trees	?

->  Decision trees use mathematical concepts like entropy, Gini impurity, and information gain to build a hierarchical structure that predicts outcomes. Each node represents a feature, branches represent decisions, and leaf nodes represent the predicted outcome. The algorithm aims to find the most effective splits in the data to create a tree that minimizes impurity and maximizes information gain

8. What is Pre-Pruning in Decision Trees	?

-> Pre-pruning, also known as early stopping, is a decision tree pruning technique where the tree's growth is halted before it fully develops, preventing overfitting by limiting its complexity. This is achieved by imposing constraints during the tree's construction, such as a maximum depth or minimum number of samples required to split a node.

9. What is Post-Pruning in Decision Trees	?

-> Post-pruning in decision trees involves trimming a fully grown tree to improve its generalization by removing branches that don't contribute significantly to accuracy on unseen data. This is done after the tree has been trained on the training data and has potentially overfitted. Post-pruning techniques like cost-complexity pruning are used to find the optimal balance between tree complexity and accuracy.

10. What is the difference between Pre-Pruning and Post-Pruning	?

-> Pre-pruning and post-pruning are two distinct methods of pruning decision trees in machine learning. Pre-pruning stops the tree from growing before it reaches its full potential, while post-pruning allows the tree to grow fully and then removes branches that don't contribute significantly to the prediction accuracy.

11. What is a Decision Tree Regressor	?

-> A Decision Tree Regressor is a supervised machine learning algorithm that predicts a continuous target variable by creating a tree-like structure of decision rules. It works by recursively splitting data based on features, ultimately leading to leaf nodes that represent the predicted numerical value.


12.  What are the advantages and disadvantages of Decision Trees	?

->  Decision trees offer several advantages, including interpretability, handling various data types, and minimal data preparation requirements. However, they also have disadvantages, such as susceptibility to overfitting, instability, and potential bias towards dominant classes in imbalanced datasets.

13.  How does a Decision Tree handle missing values	?

-> A Decision Tree handle missing values by :-

- Ignore missing values: Some Decision Tree algorithms ignore missing values during the splitting process.
- Use surrogate splits: Some algorithms use surrogate splits, which are alternative splits that can be used when a feature value is missing.
- Imputation: Replace missing values with imputed values, such as the mean or median of the feature.
- Treat missing as a separate category: Some algorithms treat missing values as a separate category, allowing the tree to learn patterns specific to missing values.

14. How does a Decision Tree handle categorical features	?

-> A Decision Tree handle categorical features by :-  

- One-hot encoding: Convert categorical features into binary features using one-hot encoding.
- Label encoding: Assign numerical values to categorical features, but this can imply an order or hierarchy.
-  Categorical splits: Some Decision Tree algorithms can handle categorical features directly, splitting on categories rather than numerical values.

15. What are some real-world applications of Decision Trees?

->  Decision trees have numerous real-world applications across various fields, including healthcare, finance, marketing, and machine learning. They are used for tasks like diagnosing diseases, assessing credit risk, predicting customer behavior, and optimizing marketing campaigns.

16. Write a Python program to train a Decision Tree Classifier on the Iris dataset and print the model accuracy

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.tree import  DecisionTreeClassifier
```

```python
# load dataset
data = load_iris()
df = pd.DataFrame(data.data, columns = data.feature_names)
df['target'] = data.target

# train-test split
x = df.drop('target', axis = 1)
y = df.target
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size= 0.2, random_state=2)

# applying Decision Tree

model = DecisionTreeClassifier()
model.fit(x_train,y_train)

y_pred = model.predict(x_test)

accuracy = accuracy_score(y_test, y_pred)

print('The accuracy score is', round(accuracy,2))
```

19. Write a Python program to train a Decision Tree Regressor on a housing dataset and evaluate using Mean
Squared Error (MSE)

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.datasets import fetch_california_housing
from sklearn.metrics import mean_squared_error

# Import libraries
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
import pandas as pd

data = fetch_california_housing()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target

x = df.drop('target', axis=1)
y = df['target']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2)

model = DecisionTreeRegressor(random_state=2)
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
# evaluation usind mean squared error
mse = mean_squared_error(y_test, y_pred)

print('The Mean Squared Error is:', round(mse, 2))

```

27. Write a Python program to train a Decision Tree Classifier, apply Cost Complexity Pruning (CCP), and
visualize its effect on accuracy

```python
clf = DecisionTreeClassifier(random_state=0)
path = clf.cost_complexity_pruning_path(x_train, y_train)
ccp_alphas = path.ccp_alphas

clfs = []
for ccp_alpha in ccp_alphas:
    clf = DecisionTreeClassifier(random_state=0, ccp_alpha=ccp_alpha)
    clf.fit(x_train, y_train)
    clfs.append(clf)

train_scores = [clf.score(x_train, y_train) for clf in clfs]
test_scores = [clf.score(x_test, y_test) for clf in clfs]

plt.figure(figsize=(8, 5))
plt.plot(ccp_alphas, train_scores, marker='o', label='Train Accuracy', color='blue')
plt.plot(ccp_alphas, test_scores, marker='o', label='Test Accuracy', color='green')
plt.xlabel("ccp_alpha (Complexity Parameter)")
plt.ylabel("Accuracy")
plt.title("Effect of CCP Pruning on Decision Tree Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
